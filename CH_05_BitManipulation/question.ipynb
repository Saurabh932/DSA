```python
'''Check Odd or Even'''
def odd_or_even(n):
    bit_mask = 1
    if n&bit_mask == 0:
        return "Even"
    return "Odd"

print(odd_or_even(5))
print(odd_or_even(22))
print(odd_or_even(38))
print(odd_or_even(35))
print(odd_or_even(3))
```

```python
'''Get ith Bit'''
def getIthBit(n, i):
    bit_mask = 1<<i
    
    if (n & bit_mask) == 0:
        return 0 
    return 1

print(getIthBit(10, 3))
```

```python
'''Set ith Bit'''
def setIthBit(n, i):
    bit_mask = 1<<i
    
    return n | bit_mask

print(setIthBit(10, 2))
```

```python
'''Clear ith Bit'''
def clearIthBit(n, i):
    bit_mask = 1<<i
    
    return n & (~bit_mask)
    
print(clearIthBit(10, 1))
```

```python
'''Update ith Bit'''
def updateIthBit(n, i, newBit):
    if newBit == 0:
        return clearIthBit(n, i)
    return setIthBit(n, i)

print(updateIthBit(10, 2, 0))
```

```python
'''Clear Bits in Range'''
def clearBitRange(n, i, j):
    a = (~0) <<(j+1)
    b = (1<<i) - 1
    
    bit_mask = a | b
    
    return  n & bit_mask


print(clearBitRange(10, 2, 4))
```

```python
'''Check If the number is Power of 2'''
def checkPow(n):
    return (n&(n-1)) == 0

print(checkPow(9))
```
