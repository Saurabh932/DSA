```python
''' 
    Max Subarray sum
'''

def maxSubarraySum(arr):
    curr_sum  = 0
    max_sum = float("-inf")
    prefix = [0]*len(arr)
    prefix[0] = arr[0]
    
    # Calculating prefix
    for i in range(1, len(arr)):
        prefix[i] = prefix[i-1] + arr[i]
        
        
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            curr_sum = prefix[j] if i==0 else prefix[j] - prefix[i-1]
            
            if max_sum < curr_sum:
                max_sum = curr_sum
                
    return max_sum

arr = [1, -2, 6, -1, 3]
print(maxSubarraySum(arr))
```

```python
def repeadMissing(grid):
    seen = set()
    ans = []
    repeated = None
    missing = None
    
    for ch in grid:
        for ele in ch:
            if ele in seen:
                repeated = ele
                
            else:
                seen.add(ele)
                
    ans.append(repeated)
    
    # missing 
    n = len(grid) * len(grid)
    
    missing = (n*(n+1)//2) - sum(seen)
    ans.append(missing)
    
    return ans
 

print(repeadMissing([[9,1,7],[8,9,2],[3,4,6]]))
```

```python
def countSetBit(num):
    count = 0
    while num>0:
        if num&1 != 0:
            count += 1
            
        num = num>>1
        
    return count

print(countSetBit(6))
```
