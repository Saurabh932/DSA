#### 1. What is a Subarray?

A subarray is:
 - A continuous part of an array
 - Elements must be from contiguous positions

#### 2. Total number of subarrays

For array of size n:

Total subarrays = 𝑛(𝑛+1)/2
	​


Example: n = 5 → 15 subarrays.

```python
def subArray(arr):
    for i in range(len(arr)):
        for j in range(i+1, len(arr)+1):
            print(arr[i:j], end=" ")
        print()
        
subArray([1,2,3,4,5])
```

#### Sum of Subarrays (Kadane’s algorithm)

```python
def sum_subarr(arr):
    for i in range(len(arr)):
        total = 0
        for j in range(i, len(arr)):
            total += arr[j]
            print(total, end=" ")
        print()
            
sum_subarr([1,2,3,4])
```
