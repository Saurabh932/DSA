#### 1. Fill empyt array and then subtrack 2

```python
def changeArr(arr, i, val):
    if i == len(arr):
        print(arr)
        return
    
    
    arr[i]=val  # Choice
    changeArr(arr, i+1, val+1) # Recursive call
    
    # Backtracking
    arr[i] -= 2
    
    return arr
    
print(changeArr([0,0,0,0,0], 0, 1))
```

#### 2. Subset

```python
def subset(s, i, ans):
    if i == len(s):
        print(ans)
        return

    # Exclude
    subset(s, i+1, ans)

    # Include
    subset(s, i+1, ans + s[i])


subset("abc", 0, "")
```

#### 3. Final Permutation

```python
def finalPermutation(s, ans):
    if len(s) == 0:
        print(ans)
        return
    
    for i in range(len(s)):
        curr = s[i]
        newS = s[:i] + s[i+1:]
        finalPermutation(newS, ans+curr)
        
print(finalPermutation('abc',""))
```

## *N-Queen*

```python
def isSafe(board, row, col):
    # Checking Vertically
    for i in range(row-1, -1, -1):
        if board[i][col] == "Q":
            return False
        
    # Checking left diagonal
    i, j = row-1, col-1
    while i>=0 and j>=0:
        if board[i][j] == "Q":
            return False
        i -= 1
        j -= 1
            
    # Checking right diagonal
    i, j = row-1, col+1
    while i>=0 and j<len(board[0]):
        if board[i][j] == "Q":
            return False
        i -= 1
        j += 1
            
    return True

def nQueens(board, row):
    
    if row == len(board):
        for r in board:
            print(" ".join(r))
        print()
        return
    
     # Choosing row to place Q
    for i in range(row):
        board[row][i] = "Q"
        nQueens(board, row+1)  # now recursively calling the function and increasing row
        board[row][i] = "O"   # Backtracking the call and removing the queen
        

row = int(input("Enter rows: "))
col = int(input("Enter columns: "))

board = [["X" for _ in range(col)] for _ in range(row)]
        
nQueens(board, 0)   
```

### N-Queen one solution

```python
def isSafe(board, row, col):
    # Check vertical
    for i in range(row-1, -1, -1):
        if board[i][col] == "Q":
            return False
        
    # Check left diagonal
    i, j = row-1, col-1
    while i >= 0 and j >= 0:
        if board[i][j] == "Q":
            return False
        i -= 1
        j -= 1
            
    # Check right diagonal
    i, j = row-1, col+1
    while i >= 0 and j < len(board[0]):
        if board[i][j] == "Q":
            return False
        i -= 1
        j += 1
            
    return True


def nQueens(board, row):
    
    if row == len(board):
        for r in board:
            print(" ".join(r))
        print()
        return True   # solution found
    
    for col in range(len(board[0])):   # FIXED
        if isSafe(board, row, col):
            board[row][col] = "Q"
            
            if nQueens(board, row+1):   # if solution found deeper
                return True             # stop everything
            
            board[row][col] = "X"       # backtrack
            
    return False


row = int(input("Enter rows: "))
col = int(input("Enter columns: "))

board = [["X" for _ in range(col)] for _ in range(row)]
        
nQueens(board, 0)
```
