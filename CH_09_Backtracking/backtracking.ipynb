#### 1. Fill empyt array and then subtrack 2

```python
def changeArr(arr, i, val):
    if i == len(arr):
        print(arr)
        return
    
    
    arr[i]=val  # Choice
    changeArr(arr, i+1, val+1) # Recursive call
    
    # Backtracking
    arr[i] -= 2
    
    return arr
    
print(changeArr([0,0,0,0,0], 0, 1))
```

#### 2. Subset

```python
def subset(s, i, ans):
    if i == len(s):
        print(ans)
        return

    # Exclude
    subset(s, i+1, ans)

    # Include
    subset(s, i+1, ans + s[i])


subset("abc", 0, "")
```

#### 3. Final Permutation

```python
def finalPermutation(s, ans):
    if len(s) == 0:
        print(ans)
        return
    
    for i in range(len(s)):
        curr = s[i]
        newS = s[:i] + s[i+1:]
        finalPermutation(newS, ans+curr)
        
print(finalPermutation('abc',""))
```

### N-Queen one solution

```python
def isSafe(board, row, col):
    # Check vertical
    for i in range(row-1, -1, -1):
        if board[i][col] == "Q":
            return False
        
    # Check left diagonal
    i, j = row-1, col-1
    while i >= 0 and j >= 0:
        if board[i][j] == "Q":
            return False
        i -= 1
        j -= 1
            
    # Check right diagonal
    i, j = row-1, col+1
    while i >= 0 and j < len(board[0]):
        if board[i][j] == "Q":
            return False
        i -= 1
        j += 1
            
    return True


def nQueens(board, row):
    
    if row == len(board):
        for r in board:
            print(" ".join(r))
        print()
        return True   # solution found
    
    for col in range(len(board[0])):   # FIXED
        if isSafe(board, row, col):
            board[row][col] = "Q"
            
            if nQueens(board, row+1):   # if solution found deeper
                return True             # stop everything
            
            board[row][col] = "X"       # backtrack
            
    return False


row = int(input("Enter rows: "))
col = int(input("Enter columns: "))

board = [["X" for _ in range(col)] for _ in range(row)]
        
nQueens(board, 0)
```

### Grid Ways

```python
def gridWays(i,j,n,m):
    # When reached last cell
    if i==n-1 and j==m-1:
        return 1
    elif i==n or j==m:  # When the pointer moves out of row or column
        return 0
    
    w1 = gridWays(i+1,j,n,m) # When moving downwards in row
    w2 = gridWays(i,j+1,n,m) # When moving rightways in column
    return w1+w2

print(gridWays(0,0,3,3))
```

### Sudoku Game Solver:

```python
def isSafe(board, row, col, num):
    # Checking row 
    for j in range(9):
        if board[row][j] == num:
            return False
        
    # Checking column
    for i in range(9):
        if board[i][col] == num:
            return False
        
        # Checking 3x3 invdividual gird and it is 9X9 board 3x3 == 1 grid and 1*9 = 9
    startRow = row - row%3
    startCol = col - col%3
    
    for i in range(3):
        for j in range(3):
            if board[startRow+i][startCol+j] == num:
                return False
    
    return True


def sudokuSolver(board):
    
    for row in range(9):
        for col in range(9):
            
            # Checking empyt cell
            if board[row][col] == 0:
                for num in range(1,10):
                    if isSafe(board, row, col, num):
                        board[row][col] = num
                        
                        # Checking if true
                        if sudokuSolver(board):
                            return True
                        
                        '''BackTrack'''
                        board[row][col] = 0
                return False
    return True


board = [
[3,0,6,5,0,8,4,0,0],
[5,2,0,0,0,0,0,0,0],
[0,8,7,0,0,0,0,3,1],
[0,0,3,0,1,0,0,8,0],
[9,0,0,8,6,3,0,0,5],
[0,5,0,0,9,0,6,0,0],
[1,3,0,0,0,0,2,5,0],
[0,0,0,0,0,0,0,7,4],
[0,0,5,2,0,6,3,0,0]
]

sudokuSolver(board)

for row in board:
    print(row)
```
