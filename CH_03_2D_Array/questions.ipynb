```python
'''
Question: Print the Sum of All Elements in a Matrix
'''

def matrixSum(matrix, row, col):
    total = 0
    
    for i in range(row):
        for j in range(col):
            total += matrix[i][j]
            
    return total


row, col = map(int, input("Enter rows and cols: ").split())

matrix = [list(map(int, input().split())) for _ in range(row)]

print("Sum:", matrixSum(matrix, row, col))
```

```python
'''
Question: Spiral Matrix
'''

def spiralMatrix(matrix, row, col):
    start_row = 0
    start_col = 0
    end_row = row - 1
    end_col = col - 1

    while start_row <= end_row and start_col <= end_col:

        # Top boundary (left → right)
        for c in range(start_col, end_col + 1):
            print(matrix[start_row][c], end=" ")
        start_row += 1

        # Right boundary (top → bottom)
        for r in range(start_row, end_row + 1):
            print(matrix[r][end_col], end=" ")
        end_col -= 1

        # Bottom boundary (right → left)
        if start_row <= end_row:
            for c in range(end_col, start_col - 1, -1):
                print(matrix[end_row][c], end=" ")
            end_row -= 1

        # Left boundary (bottom → top)
        if start_col <= end_col:
            for r in range(end_row, start_row - 1, -1):
                print(matrix[r][start_col], end=" ")
            start_col += 1

    print()


row, col = map(int, input("Enter rows and cols: ").split())
matrix = [list(map(int, input().split())) for _ in range(row)]

spiralMatrix(matrix, row, col)
```

```python
''' 
    Diagonal Sum
'''

def diagonalSum(mat, row, col):
    sum = 0
    
    # Brute force
    for i in range(row):
        for j in range(col):
            # Primary Diagonal
            if i==j:
                sum += mat[i][j]
                
            # Secondary
            elif i+j == row-1:
                sum += mat[i][j]
               
                
    # Optimized
    for i in range(row):
        sum += mat[i][i]
        
        if i != row-i-1:
            sum += mat[i][row-i-1]
    
    return sum


row, col = map(int, input("Enter rows and cols: ").split())
matrix = [list(map(int, input().split())) for _ in range(row)]

print(diagonalSum(matrix, row, col))
```

```python
'''
    Search in the sorted matrix
'''

def staircaseSearch(mat, row, col, k):
    r = 0
    c = col-1
    
    while r<row and c>=0:
        if mat[r][c] == k:
            print(f"found at ({r}, {c})")
            return True
        
        elif k < mat[r][c]:
            c -= 1
            
        else:
            r += 1
            
    print("Key not found")
    return False

row, col = map(int, input("Enter rows and cols: ").split())
k = int(input("Enter the key: "))
matrix = [list(map(int, input().split())) for _ in range(row)]

print(staircaseSearch(matrix, row, col, k))
```
