```python
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

matrix = []

for i in range(rows):
    row = list(map(int, input(f"Enter {cols} values for row {i}: ").split()))
    matrix.append(row)

print("\nMatrix is:")
for row in matrix:
    print(row)
```

```python
rows = int(input("Rows: "))
cols = int(input("Cols: "))

matrix = []

for i in range(rows):
    row = []
    for j in range(cols):
        val = int(input(f"Enter value at position ({i},{j}): "))
        row.append(val)
    matrix.append(row)

print("\nMatrix:")
for row in matrix:
    print(row)
```

```python
rows = 3
cols = 4

matrix = [[0 for j in range(cols)] for i in range(rows)]
print(matrix)
```

```python
rows, cols = map(int, input().split())

matrix = [list(map(int, input().split())) for _ in range(rows)]

print(matrix)
```
