```python
# Check Piror element is greater the current

def checkPrior(arr):
    count = 1
    max_so_far = arr[0]
    
    for i in range(1, len(arr)):
        if arr[i] > max_so_far:
            count += 1
            max_so_far = arr[i]
            
    return count
            
print(checkPrior([5, 4, 3, 2, 1]))
```

```python
# Find the balloons which are present odd number of time

def oddBalloon(arr):
    freq = {}
    for ch in arr:
        freq[ch] = freq.get(ch, 0) + 1
        
    for ch in arr:
        if freq[ch]%2 != 0:
            return ch
    return "All are even"

n = int(input())
# balloon = list(input().split(","))

balloon = ['r','g','b','b','g','y','y']
print(balloon)
print(oddBalloon(balloon))
```

```python
# Find the remaing mondkey on the tree
def remainingMonkey(n,k,j,m,p):
    count = 0
    
    # Total Number of Banana Remaining
    count += m/k
    m = m%k
    
    # Total Peanuts remaining
    count += p/j
    p = p%j
    
    # Base Case
    if (m<k and m>0):
        count += 1
    
    if (p<j and p>0):
        count += 1
        
n = int(input())
k = int(input())
j = int(input())
m = int(input())
p = int(input())
print(remainingMonkey(n,k,j,m,p))
```

```python
# Find the longest Prefix of the string in an array
def longestPrefix(arr):
    if len(arr) < 1:
        return ""
    
    res = []
    arr.sort()
    first = arr[0]
    last = arr[-1]
    
    for i in range(min(len(first), len(last))):
        if first[i] != last[i]:
            break
        res.append(first[i])
        
    return "".join(res)

n = int(input())
arr = []
for _ in range(n):
    arr.append(input())
    
print(longestPrefix(arr))
```

```python
# Count number of sundays in n days when started with any day
def sundayCount(n, day):
    days = {'mon': 6, 'tue':5, 'wed':4, 'thu':3, 'fri':2, 'sat':1, 'sun':0}
    
    count = 0
    
    if n - days[day] > 1:
        count = 1 + (n-days[day])/7
        
    return int(count)

n = int(input())
day = input()
print(sundayCount(n, day))
```

```python
# Find which row has maximum 1
def countOne(mat, r, c):
    row = 0
    max_one = 0
    for r in range(r):
        current = 0
        current_row = 0
        for c in range(c):
            if mat[r][c] == 1:
                current += 1
            current_row = r
        
        if current > max_one:
            max_one = current
            row = current_row
            
    if max_one == 0:
        return -1
    return row+1

            
print(countOne([[0,1,0],[1,0,1],[1,0,1]], 3, 3))
```

```python
def sumItoJ(i, j):
    sum = 0
    for x in range(i, j+1):
        sum += x
        
    return sum

i = int(input())
j = int(input())
print(sumItoJ(i, j))
```
