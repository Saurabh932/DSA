#### If size of array is given 

```python
n = int(input("Enter the size: "))
arr = list(map(int, input().split()))
for i in range(n):
    print(arr[i], end=' ')
```

```python
n = int(input("Enter the size: "))
arr = list(map(int, input("Enter the elements: ").split()))
print(arr)
```

#### Dynamic Array

```python
n = int(input("Enter the size: "))
arr = list(map(int, input().split()))

# Printing the elements using the simple loop
for i in range(n):
    print(arr[i], end=" ")
print()
  
# Printing the elements using ehanced loop
for ele in arr:
    print(ele, end=" ")
```

#### If size of array is not giveen and take space sperated values


```python
n = int(input("Enter the size: "))
arr = list(map(int, input().strip().split(", ")))
print(arr)
```
