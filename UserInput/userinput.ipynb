#### If size of array is given 

1. Normal Array

```python
n = int(input("Enter the size: "))
arr = list(map(int, input().split()))

for i in range(n):
    print(arr[i], end=' ')
```

2.  Dynamic Array

```python
n = int(input("Enter the size: "))
arr = list(map(int, input().split()))

# Printing the elements using the simple loop
for i in range(n):
    print(arr[i], end=" ")
print()
  
# Printing the elements using ehanced loop
for ele in arr:
    print(ele, end=" ")
```

#### If size of array is not given and take space and comma sperated values


1. Using String

```python
input_str = input()
# nums = input_str.split(" ")  # Space sperated
nums = input_str.split(",")  # Comma sperated


arr = [int(num) for num in nums]

print(" ".join(map(str, arr)))
```

2. Using list

```python
n = int(input("Enter the size: "))
arr = list(map(int, input().split(",")))
print(arr)
```

```python
n = int(input("Enter the size: "))
arr = list(map(int, input("Enter the elements: ").split()))
print(arr)
```

### If starting and ending with brackets
arr = [1,2,3,4,5]

```python
input_str = input()
# nums = input_str.split(" ")  # Space sperated

# Starting and ending with brackets
if input_str.startswith("[") and input_str.endswith["]"]:
    input_str = input_str[1:-1]
    
nums = input_str.split(",")  # Comma sperated

arr = [int(num) for num in nums]

print(" ".join(map(str, arr)))
```

```python
def validString(s):
    # freq = {}
    
    # for ch in s:
    #     freq[ch] = freq.get(ch, 0) + 1
        
    # if freq.get('*', 0) == freq.get('#', 0):
    #     return 'Valid String'
    # return "Not Valid String"
    
    count = 0
    for ch in s:
        if ch == '*':
            count += 1
        else:
            count -= 1
            
    return count
     
    
print(validString("***##**####"))
```

```python
# Check Piror element is greater the current

def checkPrior(arr):
    count = 1
    max_so_far = arr[0]
    
    for i in range(1, len(arr)):
        if arr[i] > max_so_far:
            count += 1
            max_so_far = arr[i]
            
    return count
            
print(checkPrior([5, 4, 3, 2, 1]))
```
