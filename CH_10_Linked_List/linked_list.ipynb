```python
'''Node'''
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
'''LinkedList'''
class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0
        
    '''Inserting at Start'''
    def insert_at_start(self, data):
        new_node = Node(data)
        self.size += 1
        
        if self.head is None:
            self.head = self.tail = new_node
            return
        
        new_node.next = self.head
        self.head = new_node
        
        
    '''Inserting at end'''
    def insert_at_end(self, data):
        new_node = Node(data)
        self.size += 1
        
        if self.head is None:
            self.head = self.tail = new_node
            return
        
        self.tail.next = new_node
        self.tail = new_node
        
        
    '''Insertin at x-position'''
    def insert_at_position(self, idx, data):        
        if idx == 0 or self.head is None:
            self.insert_at_start(data)
            return
        
        new_node = Node(data)
        self.size += 1
        
        temp = self.head
        count = 0
        
        while temp.next and count < idx-1:
            temp = temp.next
            count += 1
        
        # Inserting Node 
        new_node.next = temp.next
        temp.next = new_node
                
        # Update tail if inserted at end
        if new_node.next is None:
            self.tail = new_node
                
                
    '''Displaying Linked List'''
    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("NULL")
        
        
        
    '''Searching in LL uisng loops'''
    def searchNode(self, target):
        # When size is 0
        if self.size == 0:
            print("The Linked LIst is empty")
            return -1
        
        val = self.head
        for i in range(self.size):
            if val.data == target:
                # print(i)
                return i
            
            val = val.next
            
            
    
    '''Recursive Search in LinkedList'''
    def helper(self, head, target):
        if head is None:
            return -1
        
        if head.data == target:
            return 0
        
        idx = self.helper(head.next, target)
        if idx == -1:
            return -1
        
        return idx+1
        
        
    def recSearch(self, target):
        return self.helper(self.head, target)
        
        
    '''Removing First/Head element'''
    def firstRemove(self):
        # When size is 0
        if self.size == 0:
            print("The linked list is empty")
            return f"-inf"
        
        # When size is 1
        if self.size == 1:
            val = self.head.data
            self.head = self.tail = None
            self.size = 0
            return val
            
        
        val = self.head.data
        self.head = self.head.next
        
        self.size -= 1
        return val
    
    
    '''Removing Tail/Last Elements'''
    def lastRemove(self):
        if self.size == 0:
            print("The Linked List is empty")
            return f"-inf"
        
        if self.size == 1:
            val = self.tail.data
            self.head = self.tail = None
            self.size = 0
            return val
        
        prev = self.head
        for i in range(self.size-2):
            prev = prev.next
        
        
        val = prev.next.data
        prev.next = None
        self.tail = prev
        
        self.size -= 1
        return val
    
    
    '''Reverse a Linked List'''
    def reverseLL(self):
        prev = None
        curr = self.tail = self.head
        next = None
        
        while curr != None:
            next = curr.next
            curr.next = prev 
            prev = curr
            curr = next
            
        self.head = prev
        
        
    
    '''Find and deleth nth node from end'''
    def deleteNthFromEnd(self, target):
        # Calculating size
        size = 0
        temp = self.head
        while temp is not None:
            temp = temp.next
            size+=1
              
        # If deleting the first node
        if target == size:
            self.head = self.head.next # Remove first
            return
        
        # Find node just before the one to delete
        i = 1
        iToFind = size-target
        prev = self.head
        while i < iToFind:
            prev = prev.next
            i+=1
    
        # Delete the node
        prev.next = prev.next.next
        return
    
    
    '''Check if LL is Palindrome'''
    # Helper function - Slow-Fast Apporach (Floyds Finding Algorithm)
    def findMid(self):
        slow = self.head
        fast = self.head
        
        while fast != None and fast.next != None:
            slow = slow.next  # +1
            fast = fast.next.next  # +2
            
        return slow  # Mid
    
    
    def checkPalindrome(self):
        # Base Case
        if self.head == None or self.head.next == None:
            return True
        
        # Step 1 - Find Mid
        midNode = self.findMid()
        
        # Step 2 - Reverse 2nd half
        prev = None
        curr = midNode
        
        while curr != None:
            next = curr.next
            curr.next = prev
            prev = curr
            curr = next
            
        right = prev   # right half head
        left = self.head
        
        # Step 3 - Check left and right half
        while right != None:
            if left.data != right.data:
                return False
            
            left = left.next
            right = right.next

        return True
    

    '''Detect Loop or Cycle in LL'''
    def isCycle(self):
        slow = self.head
        fast = self.head
        
        while fast != None and fast.next != None:
            slow = slow.next
            fast = fast.next.next

            if slow == fast:
                return True
        return False
            
            
ll = LinkedList()

# ll.insert_at_start(10)
# ll.insert_at_start(5)
# ll.insert_at_end(20)
# ll.insert_at_end(30)

# ll.insert_at_position(2, 47)   # inserting in the middle

# ll.display()
# print()
# print(ll.size)


# print(f"The target is at",  {ll.searchNode(47)})
# print(f"The target is at",  {ll.recSearch(20)})

# ll.firstRemove()
# ll.display()

# ll.lastRemove()
# ll.display()

# ll.reverseLL()
# ll.display()

# ll.deleteNthFromEnd(5)
# ll.display()


# ll.insert_at_end(1)
# ll.insert_at_end(2)
# ll.insert_at_end(2)
# ll.insert_at_end(1)

# ll.display()

# print("Check Palindrome: ", ll.checkPalindrome())


ll.insert_at_end(1)
ll.insert_at_end(2)
ll.insert_at_end(3)
ll.insert_at_end(1)

ll.display()

print("Is loop present: ", ll.isCycle())
```
