```python
'''Node'''
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
'''LinkedList'''
class LinkedLikst:
    def __init__(self):
        self.head = None
        self.tail = None
        
    '''Inserting at Start'''
    def insert_at_start(self, data):
        new_node = Node(data)
        
        if self.head is None:
            self.head = self.tail = new_node
            return
        
        new_node.next = self.head
        self.head = new_node
        
        
    '''Inserting at end'''
    def insert_at_end(self, data):
        new_node = Node(data)
        
        if self.head is None:
            self.head = self.tail = new_node
            return
        
        self.tail.next = new_node
        self.tail = new_node
        
        
    '''Insertin at x-position'''
    def insert_at_position(self, idx, data):
        new_node = Node(data)
        
        if idx == 0 or head is None:
            self.insert_at_start(data)
            return
        
        new_node = Node(data)

        temp = self.head
        count = 0
        
        while temp.next and count < idx-1:
            temp = temp.next
            count += 1
            
            if temp.next is None:
                self.tail.next = new_node
                self.tail = new_node
                
            else:
                new_node.next = temp.next
                temp.next = new_node
                
                
    '''Displaying Linked List'''
    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
            
ll = LinkedLikst()

ll.insert_at_start(10)
ll.insert_at_start(5)
ll.insert_at_end(20)
ll.insert_at_end(30)

ll.insert_at_position(2, 47)   # inserting in the middle

ll.display()
```
