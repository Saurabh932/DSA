### Merge Sort

```python
def mergeSort(arr, s, e):
    if s>=e :
        return
    
    mid = s+(e-s)//2
    
    # Sorting left and right halfs
    mergeSort(arr, s, mid)
    mergeSort(arr, mid+1, e)
    
    # Merging all the parts
    merge(arr, s, mid, e)
    
    
def merge(arr, s, mid, e):
    temp = [0]*(e-s+1)
    i = s   # Iterating in 1st half
    j = mid+1  # Iterating in 2nd half
    k = 0  # Iterating in temp
    
    while i<=mid and j<=e:
        if arr[i] <= arr[j]:
            temp[k] = arr[i]
            i+=1
        else:
            temp[k] = arr[j]
            j+=1
        k+=1
        
    # For leftover elements in 1st half
    while i<=mid:
        temp[k]=arr[i]
        i+=1
        k+=1
        
    
    # For leftover elements in 2nd half
    while j<=e:
        temp[k]=arr[j]
        j+=1
        k+=1
        
    # Copy temp back into original arr
    for i in range(len(temp)):
        arr[s+i] = temp[i]
        
        
n = int(input("Enter the size of array: "))
arr = list(map(int, input("Enter the elements: ").split()))
mergeSort(arr, 0, n-1)
for i in range(n):
    print(arr[i], end=" ")
```

### Quick Sort

```python
def partition(arr, s, e):
    pivot = arr[e]
    i = s-1
    
    for j in range(s, e):
        if arr[j] < pivot:
            i+=1
            
            # Swapping
            arr[i], arr[j] = arr[j], arr[i]
            
    i+=1
    arr[i], arr[e] = arr[e], arr[i]
    return i


def quickSort(arr, s, e):
    if s>=e:
        return
    
    # Partition index
    partitionIdx = partition(arr, s, e)
    
    # Left and Right half
    quickSort(arr, s, partitionIdx-1)
    quickSort(arr, partitionIdx+1, e)
    

n = int(input())
arr = list(map(int, input().split()))

quickSort(arr, 0, n-1)

for i in range(n):
    print(arr[i], end=" ")
```

```python
# Search in rotated sorted array

def rotatedArray(arr, target, s, e):
    if s>e:
        return -1
        
    mid = s + (e-s)//2
    
    if arr[mid] == target:
        return mid
    
    # Case 1: If the mid is on Line 1
    if (arr[s] < arr[mid]):
        # Case a: left
        if arr[s] <= target <= arr[mid]:
            return rotatedArray(arr, target, s, mid-1)
        
        # Case b: Right
        else:
            return rotatedArray(arr, target, mid+1, e)

    # Case 2: if the mid is on line 2
    else:
        # Case c: Right
        if arr[mid] <= target <= arr[e]:
            return rotatedArray(arr, target, mid+1, e)
        
        # Case d: Left
        else:
            return rotatedArray(arr, target, s, mid-1)
        
        
n = int(input())
arr = list(map(int, input().split()))
target = int(input())

print(rotatedArray(arr, target, 0, n-1))
```
